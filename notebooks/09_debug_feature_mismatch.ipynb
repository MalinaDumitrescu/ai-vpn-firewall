```python
import pandas as pd
import numpy as np
import xgboost as xgb
import json
from src.utils.paths import load_paths

paths = load_paths()

model_path = paths.repo_root / "artifacts" / "xgb" / "model.json"
booster = xgb.Booster()
booster.load_model(str(model_path))

importance = booster.get_score(importance_type="gain")
sorted_imp = sorted(importance.items(), key=lambda x: x[1], reverse=True)

print("\n--- Top 20 Features by Gain ---")
for f, score in sorted_imp[:20]:
    print(f"{f:<30} : {score:.2f}")

bad_features = ["f_duration_s", "f_bytes_per_s", "f_pkts_per_s", "f_total_bytes", "f_total_pkts"]
print("\n--- Suspicious Features Rank ---")
for bf in bad_features:
    rank = next((i for i, (f, s) in enumerate(sorted_imp) if f == bf), -1)
    if rank != -1:
        print(f"{bf:<20} is rank #{rank+1} (score: {sorted_imp[rank][1]:.2f})")
    else:
        print(f"{bf:<20} not found in model features")
```
